```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
```

```python
df=pd.read_csv('pokemon_data.csv')
df.head()
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
df=df.drop('Unnamed: 0',axis=1)
```

```python
df.head()
```

```python
le=LabelEncoder()
df['type1']=le.fit_transform(df['type1'])
df['type2']=le.fit_transform(df['type2'])
df['classfication']=le.fit_transform(df['classfication'])
```

```python
x=df.drop('is_legendary', axis=1)
y=df['is_legendary']
```

```python
df.head()
```

```python
x['capture_rate'] = x['capture_rate'].astype(str).str.extract('(\\d+)').astype(float)
x['capture_rate'] = x['capture_rate'].fillna(x['capture_rate'].mean())
x=StandardScaler().fit_transform(x)
```

```python
x_train,x_test,y_train,y_test= train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
import keras
from keras.models import Sequential
from keras.layers import Dense
from keras.layers import Dropout
from sklearn.metrics import accuracy_score,confusion_matrix
```

```python
model=Sequential()
#input layer
model.add(Dense(6,kernel_initializer='he_uniform',activation='relu',input_dim=12))
#Dropout Layer
model.add(Dropout(0.3))
#Hidden Layer
model.add(Dense(6,kernel_initializer='he_uniform',activation='relu'))
#Dropout layer
model.add(Dropout(0.3))
#Output Layer
model.add(Dense(1,kernel_initializer='glorot_uniform',activation='sigmoid'))
print("Model Built Sucessfully")
```

```python
model.compile(optimizer='Adamax', loss='binary_crossentropy', metrics=['accuracy'])
print("Compilation Successful")
```

```python
t_model=model.fit(x_train,y_train,batch_size=10,epochs=100)
```

```python
model.summary()
```

```python
preds=model.predict(x_test)
```

```python
preds
```

```python
accuracy_score(y_test,preds.round())
```

```python
confusion_matrix(y_test,preds.round())
```
