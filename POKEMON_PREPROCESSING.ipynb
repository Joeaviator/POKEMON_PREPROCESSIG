```python
df=df.drop('Unnamed: 0',axis=1)
```

```python
df.head()
```

```python
le=LabelEncoder()
df['type1']=le.fit_transform(df['type1'])
df['type2']=le.fit_transform(df['type2'])
df['classfication']=le.fit_transform(df['classfication'])
```

```python
x=df.drop('is_legendary', axis=1)
y=df['is_legendary']
```

```python
df.head()
```

```python
x['capture_rate'] = x['capture_rate'].astype(str).str.extract('(\\d+)').astype(float)
x['capture_rate'] = x['capture_rate'].fillna(x['capture_rate'].mean())
x=StandardScaler().fit_transform(x)
```

```python
x_train,x_test,y_train,y_test= train_test_split(x,y,test_size=0.2,random_state=42)
```
